```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from sklearn.preprocessing import LabelEncoder, StandardScaler
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("customer_attrition.csv")
```

```python
df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis = 1)
```

```python
encoder = LabelEncoder()
df['Geography'] = encoder.fit_transform(df['Geography'])
df['Gender'] = encoder.fit_transform(df['Gender'])
```

```python
scaler = StandardScaler()
df['Balance'] = scaler.fit_transform(df['Balance'].values.reshape(-1, 1))
df['EstimatedSalary'] = scaler.fit_transform(df['EstimatedSalary'].values.reshape(-1, 1))
```

```python
X = df.drop(['Exited'], axis = 1)
y = df['Exited']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15)
```

```python
num_features = 10
num_samples = 10000
lr_list = [0.1, 0.01, 0.001]
epoch_list = [10, 20, 30]
batch_size = 100
```

```python
model1 = Sequential([
    Input(shape=(num_features,)),
    Dense(128, activation='relu'),
    Dense(1, activation='sigmoid')
])
model1.summary()
```

```python
for lr in lr_list:
    for epoch in epoch_list:
        model1.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr), loss=tf.keras.losses.BinaryCrossentropy(), metrics=['Accuracy'])
        h = model1.fit(X_train, y_train, epochs=epoch, batch_size=batch_size, validation_split=0.15)
        plt.plot(h.history['loss'], label='training loss')
        plt.plot(h.history['val_loss'], label='validation loss')
        plt.title(f'MODEL 1, learning rate:{lr}, epochs:{epoch}')
        plt.legend()
        plt.show()
```

```python
model2 = Sequential([
    Input(shape=(num_features,)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])
model2.summary()
```

```python
for lr in lr_list:
    for epoch in epoch_list:
        model2.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr), loss=tf.keras.losses.BinaryCrossentropy(), metrics=['Accuracy'])
        h = model2.fit(X_train, y_train, epochs=epoch, batch_size=batch_size, validation_split=0.15)
        plt.plot(h.history['loss'], label='training loss')
        plt.plot(h.history['val_loss'], label='validation loss')
        plt.title(f'MODEL 2, learning rate:{lr}, epochs:{epoch}')
        plt.legend()
        plt.show()
```

```python
model3 = Sequential([
    Input(shape=(num_features,)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
model3.summary()
```

```python
for lr in lr_list:
    for epoch in epoch_list:
        model3.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr), loss=tf.keras.losses.BinaryCrossentropy(), metrics=['Accuracy'])
        h = model3.fit(X_train, y_train, epochs=epoch, batch_size=batch_size, validation_split=0.15)
        plt.plot(h.history['loss'], label='training loss')
        plt.plot(h.history['val_loss'], label='validation loss')
        plt.title(f'MODEL 3, learning rate:{lr}, epochs:{epoch}')
        plt.legend()
        plt.show()
```

All hyperparameters tested seem to converge at the same train and val accuracy.
