```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("/home/mca/Documents/220968280/week1/bodyfat.csv")
```

```python
df
```

```python
num_samples = 252
num_features = 14
X = df.drop(["BodyFat"], axis=1)
y = df['BodyFat']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = Sequential([
    Input(shape=(num_features,), name='input'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='relu')
])
```

```python
model.summary()
```

Model summary: Shows our 3 hidden layers of 128, 64 and 32 neurons each with a single neuron in the output layer. Number of parameters in each hidden layer is 1920, 8256 and 2080 respectively. We are updating all 12289 parameters here, which is why we have 0 non-trainable parameters.

```python
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.1), loss='MeanSquaredError')
```

```python
epochs = 50
batch_size = 32
h = model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, validation_split=0.1)
```

```python
loss = model.evaluate(X_test, y_test, batch_size=batch_size)
print("Test Loss:", loss)
```
