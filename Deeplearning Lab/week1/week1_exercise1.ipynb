```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import tensorflow as tf
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("iris.csv", header=None)
```

```python
df.columns = ['slength', 'swidth', 'plength', 'pwidth', 'species']
```

```python
df
```

```python
encoder = LabelEncoder()
df['species'] = encoder.fit_transform(df['species'])
```

```python
X = df.drop(['species'], axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
```

```python
num_features = 4
num_samples = 150
```

```python
model = Sequential([
    Input(shape=(num_features,), name='input'),
    Dense(8, activation='relu'),
    Dense(4, activation='relu'),
    Dense(3, activation='softmax')
])
```

```python
model.summary()
```

Model summary: Shows our 2 hidden layers of 8, 4 neurons each with 3 neurons in the output layer. Number of parameters in each hidden layer is 40, 36 respectively. We are updating all 91 parameters here, which is why we have 0 non-trainable parameters.

```python
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.1), loss=tf.keras.losses.SparseCategoricalCrossentropy(), metrics=['accuracy'])
```

```python
epochs = 100
batch_size = 4
model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, validation_split=0.1)
```

```python
loss, accuracy = model.evaluate(X_test, y_test, batch_size=batch_size)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
